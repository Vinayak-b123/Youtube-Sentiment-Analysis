```python
data_2 = pd.read_csv("data/YoutubeCommentsDataSet.csv")
```

```python
category = []

for x in data_2["Sentiment"]:
    if (x=='neutral'):
        category.append(1)
    elif (x=='negative'):
        category.append(0)
    elif (x=='positive'):
        category.append(2)

data_2['category'] = category
```

```python
data_2.head()
```

```python
data_2.drop(["Sentiment"], axis=1, inplace=True)
```

```python
data_2["Comment"] = data_2["Comment"].str.strip()
```

```python
data_2['Comment'] = data_2['Comment'].str.replace(r'http\S+|www\S+', '', regex=True)
data_2['Comment'] = data_2['Comment'].str.replace(r'[\n\r\t]', ' ', regex=True)
```

```python
data_2 = data_2.rename(columns={"Comment":"clean_comment"})
```

```python
complete_data = pd.concat([data_1, data_2], axis=0).reset_index(drop=True)
```

```python
complete_data["clean_comment"].duplicated().sum()
```

```python
complete_data.drop_duplicates(inplace=True)
complete_data["clean_comment"].duplicated().sum()
```

```python
complete_data.to_csv("data/clean_comments.csv")
```

```python
complete_data['clean_comment'] = (
    complete_data['clean_comment']
    .str.replace(r'http\S+|www\S+', '', regex=True)   # URLs
    .str.replace(r'[^\x00-\x7F]+', ' ', regex=True)   # non-English
    .str.replace(r'\s+', ' ', regex=True)             # normalize spaces
    .str.strip()
)
```
